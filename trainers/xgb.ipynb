```python
import time
import joblib
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score
```

```python
best_xgb, best_score_xgb, best_X_train_xgb, best_X_test_xgb, best_y_train_xgb, best_y_test_xgb, best_predict_xgb = joblib.load("../bestxgb.pkl")
X_scaled, y = joblib.load("../xy.pkl")
count = 0

while count < 1000000:
  with open('done.txt', 'r') as f:
    c = f.read().strip().lower()
    if c == "true" or c == "done":
      break

    X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=.3)

    model = XGBClassifier(booster='gblinear', reg_lambda=2, alpha=0.01, eta=0.8, objective="binary:logistic", eval_metric="error", n_estimators=200)
      
    model.fit(X_train, y_train)
    predict = model.predict(X_test)
  
    print("Iter", count, "Best: ", best_score_xgb)
    print("Testing accuracy: ", accuracy_score(y_test, predict))
    score = f1_score(y_test, predict)
    print("F1 score: ", score)
    if score > best_score_xgb:
      print("New Best Score")
      best_xgb = model
      best_score_xgb = score
      best_X_train_xgb = X_train
      best_X_test_xgb = X_test
      best_y_train_xgb = y_train
      best_y_test_xgb = y_test
      best_predict_xgb = predict
  count += 1
  time.sleep(1)
```
